매개변수 기본값 (Default Parameter)

```javascript
function add(num1, num2 = 10) {
    return num1+num2;
}

add(20);
```

arguments 객체

```javascript
//함수 호출시 사용한 인수 목록
function sum() {
    let total = 0;
    for (const arg of arguments){
        total += arg;
    }

    return total;
}
```

나머지 매개변수 (Rest Parameter, 가변 매개변수)

```javascript
// ESG+
// ... : 가변 매개변수
function sum(num1,...params){
    console.log(num1,params);
}

sum(10,20,30,40)
```

```javascript
// 전개 연산자
function add(num1, num2) {
    return num1 + num2
}

const nums = [10,20];

console.log(add(nums[0],nums[1]));
console.log(add(...nums));
```

```javascript
const nums1 = [10,20,30,40];
const nums2 = [...nums1];
console.log(nums2, nums1 === nums2);
```

```javascript
const person = {name: "김철수", age:20};
const person2 = {...person};

console.log(person2, person === person2);
```

#### 비구조화 할당
- 구조 -> 비구조화(구조를 분해) -> 변수에 할당


```javascript
const nums = [10,20,30,40,50];

const[a,b,c] = nums;

console.log(a,b);

```

```javascript
const [a, , c] = nums;
console.log(a,c);
```

```javascript
const [a,b, ...rest] = nums;
console.log(a,b,rest);
```

```javascript
const person = {name:'김철수', age:20};

const name = person.name;
const age = person.age;
console.log(name, age);
```

```javascript
const {name: aa, age: bb} = person;

console.log(aa,bb);
```

#### 4. 값으로서의 함수 (일급 객체)와 콜백

#### 4.1 용어의 유래: 차별 없는 '동등한 권리’

#### 4.2 일급 객체로서의 3대 속성
- 변수 할당 (Assignment)
- 매개변수 전달 (Argument Passing) / 콜백 함수 / 사용자 정의 함수 / 일회용 함수
- 반환값 활용 (Return Value)

```javascript
function func1() {
    console.log("func1()");
}

function func2(f) {
    console.log("func2()");
    f();
}

func2(func1());
```

```javascript
function square(num){
    return num*num*num;
}

function calc(nums,callback) {
    let total = 0;
    for (const num of nums) {
        if (typeof callback === 'function') {
            total += callback(num);
        }
    }

    return total;
}
```

```javascript
calc([10,20,30], square);
```

```javascript
calc ([10,20,30], function(num){
    return num * num;
});
```

```javascript
calc([10,20,30], x => x*x);
```

### 5. 화살표 함수 (Arrow Function)

#### 5.1 정의 방법 및 스코프 생략 규칙

```javascript
- function 키워드 생략
- 매개변수 저의부와 코드 실행 영역({}) 사이에 => 로 표기
```

```javascript
function square(num) {
    return num*num;
}
```

```javascript
const square2 = (num) => {
    return num * num;
};

square2(10);
```

```javascript
//const square3 = (num) => num*num;
//const square3 = (x) => x*x;
const square3 = x => x*x;

square3(10);
```
