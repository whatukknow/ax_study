#### 6.2 선택문 (switch - case)

````javascript
const rank = 1;

switch(rank) {
    case 1:
        console.log("금메달");
        break;
    case 2:
        console.log("은메달");
        break;
    case 3:
        console.log("동메달");
        break;
    default:
        console.log("등수는 1,2,3 중에서 입력")
}
````

- switch ~ case: case 키워드에 매칭이 되면 그게 바로 코드 실행 시작점이고, break를 만날때까지 실행
- break: 실행흐름을 중단
- default: case에 매칭된 키워드가 없으면 마지막에 유입되는 구간

````javascript
/*
const grade = 1 ~ 6;
1 - 1학년 6 - 6학년, 1~6사이 값이 아니면 잘못 입력 출력
*/

const grade = 3;

switch(grade){
    case 1: console.log("1학년"); break;
    case 2: console.log("2학년"); break;
    case 3: console.log("3학년"); break;
    case 4: console.log("4학년"); break;
    case 5: console.log("5학년"); break;
    case 6: console.log("6학년"); break;
    default: console.log("잘못 입력");
}


````
