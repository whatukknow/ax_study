## 2장 1강 : System·User·Assistant 메시지 구조

### 3. 역할별 메시지 템플릿 조립하기

- System: 모델의 역할(페르소나), 상황, 과업, 규칙 정의, 최우선순위 / SystemMessage
- User: 실제 질의 응답에 대한 사용자 질문, 처리할 데이터의 전달, 가변적, 프론프트 템플릿 / HumanMessage
- Assistant: AI가 출력한 응답 / AIMessage

- User - Assistant -> 대화의 맥락 정보를 쌍으로 전달, 1쌍 - 턴, 여러쌍 -> 멀티 턴
- BassMessage의 하위 클래스(HumanMessage, SystemMessage, AIMessage, ToolMessage)

#### 3.1 라이브러리 로드 및 환경 설정

```python
from langchain.chat_models import init_chat_model
from langchain_core.output_parsers import StrOutputParser
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage, BaseMessage

model = init_chat_model(model_provider="ollama", model="mistral")

parser = StrOutputParser()

chain = model | parser
```

#### 3.2 역할별 메시지 템플릿 생성 및 실행

```python
from langchain_core.prompts import ChatPromptTemplate

prompt_template = ChatPromptTemplate.from_messages([
    ("system", "당신은 재치있게 컴퓨터의 개념을 설명하는 유치원 교사입니다."),
    ("user", "다음 개념을 설명하세요: {text}")
])

chain2 = prompt_template | model | parser

response = chain2.invoke({"text": "함수"})

response
"""
' 안녕하세요! 컴퓨터 교육 강사입니다. 오늘은 함수(Function)에 대해 설명해 보도록 합니다.\n\n함수는 특정 작업을 수행하는 것을 말합니다. 컴퓨터에서는 함수를 사용하여 복잡한 문제를 단순하게 분할하여 해결할 수 있습니다. 예를 들어, 계산 문제를 풀기 위해 다음과 같은 함수를 만들 수 있습니다.\n\n1. 추가(Addition) 함수: 두 수를 입력받아 두 수의 합을 반환합니다.\n2. 뺄셈(Subtraction) 함수: 두 수를 입력받아 뺄 수를 입력받아 두 수의 차를 반환합니다.\n3. 곱셈(Multiplication) 함수: 두 수를 입력받아 두 수의 곱을 반환합니다.\n4. 나누기(Division) 함수: 두 수를 입력받아 나누어야 할 수를 입력받아 두 수의 몫을 반환합니다.\n\n함수는 이러한 작업을 수행하는 프로그램을 만드는 데 중요한 역할을 합니다. 이러한 함수를 사용하여 복잡한 프로그램을 작성하는 것은 많은 작업을 효율적으로 하는 데 도움이 됩니다.\n\n이제 당신은 함수의 개념을 이해하셨습니다. 함수를 사용하여 컴퓨터 프로그래밍에서 복잡한 문제를 효율적으로 풀어볼 수 있습니다. 함수를 사용하는 방법에 대해서도 알아보기 바랍니다!'
"""


```

### 4. 대화 이력(Memory) 구조 설계하기

#### 4.1 Assistant 메시지 기반 문맥 유지

```python
messages: list[BaseMessage] = [
    HumanMessage("내 이름 김철수야. 만나서 반가워"),
    AIMessage("반갑습니다. 철수님, 무엇을 도와드릴까요?"),
    HumanMessage("오늘 점심 메뉴를 추천해줘"),
    AIMessage("오늘 점심은 돈까스가 좋을 것 같아요."),
    HumanMessage("오늘 점심 추천이 뭐였지?")
    #HumanMessage("내 이름이 뭐라고 했지?")
]

response = chain.invoke(messages)
response
```
