```python
import re
from langchain_core.messages import HumanMessage
from langchain_core.runnables import chain
from langchain_core.output_parsers import StrOutputParser
from lib import get_model


@chain
def protected_message(message: str)-> list[HumanMessage]:
    pattern = re.compile(r"(010)\D*(\d{4})\D*(\d{4})")

    message = pattern.sub(r"\g<1>-****-****",message)

    return [HumanMessage(message)]


model = get_model()
parser = StrOutputParser()

chain2 = protected_message | model | parser
```

```python
chain2.invoke("나의 전화번호는 010-0000-0000")
```

```python
for chunk in chain2.stream("나의 전화번호는 010-0000-0000 입니다. 당신의 번호는?"):
    print(chunk, end="")
```

```python
messages = [
    "나의 전화번호는 010-0000-0000 입니다. 당신의 번호는?",
    "안녕하세요",
    "반갑습니다."
]

for i, res in enumerate(chain2.batch(messages)):
    print(i, res)
```

```python
async def async_messages():
    async for chunk in chain2.astream("나의 전화번호는 010-0000-0000 입니다. 당신의 번호는?"):
        print(chunk, end="")
```

```python
await async_messages()
```
