```python
from langchain_core.prompts import ChatPromptTemplate

prompt_template = ChatPromptTemplate.from_messages([
    (
        "system",
        "당신은 친절한 프리랜서 영어 강사입니다. "
        "가장 자연스러운 영어 표현 1개와 그 이유를 설명해 주세요."
    ),
    (
        "user",
        "{user_question}"
    )
])
```

```python
sample_prompt = prompt_template.invoke({
    "user_question": "식당에서 남은 음식을 포장해 달라고 할 때 어떻게 말하나요?"
})

print(sample_prompt)
```
