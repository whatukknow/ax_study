1. 1세대: % 연산자 방식

```python
message = "나의 이름은 %3s입니다. 나이는 %d살입니다."

message % ("홍길동", 20)
```

```python
message = "주문번호: %05d" % 10
message
```

```python
#message = "숫자: %f" % 10.123456789
#message = "숫자: %.2f" % 10.123456789
message = "숫자: %010.2f" % 10.123456789
message
```

```python
message = "숫자: %d" % 10.12345
message
```

2. 2세대: str.format() 매세드 방식

```python
##message = "내 이름 {0}이고, 나이는 {1}입니다"
message = "내 이름 {0}이고, 나이는 {1}입니다"
message.format("철수", 20)
```

```python
message = "내 이름 {name}이고, 나이는 {age}입니다"
message.format(name="철수", age=20)


```

message = "내 이름 {0}이고, 나이는 {1}입니다"

```python
message = "내 이름 {name}이고, 나이는 {0}입니다"
message.format(20, name="철수") #항상 숫자로 된 부분이 먼저 와야 한다
```

3. 3세대 f-string 방식

```python
name = "영희"
age = 20

message = f"내이름은 {name}이고, 나이는 {age}입니다."

message
```
