## 11장 3강 비동기 태스크 동시 처리 및 실무 실습

1. asyncio.gather()


```python
- 동시에 실행, 마지막 작업까지 대기했다가 한꺼번에 결과를 반환해줌.
- 
```

```python
import asyncio

async def work(title, seconds):
    await asyncio.sleep(seconds)
    print(title, "작업")
    return title + "완료!"
```

```python
async def main():
    await work("작업A", 2)
    await work("작업B", 3)
    await work("작업C", 1)
    
```

```python
async def main():
    results = await asyncio.gather(work("작업A",2), work("작업B", 3), work("작업C",1))
    print(results)
```

```python
await main()
```
