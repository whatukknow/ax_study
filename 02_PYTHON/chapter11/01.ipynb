## 동기(Sync) vs 비동기(Async)의 이해

#동기식 함수


```python
import time

def work(title, seconds):
    time.sleep(seconds)
    print(title, "완료!")


start = time.perf_counter() #작업 시작 시간
work("A작업",2)
work("B작업",3)
work("C작업",1)

end = time.perf_counter() #작업 종료 시간

print("걸린시간:", end - start)

```

```python
# 비동기 함수
import asyncio
import time

async def async_work(title, seconds):

    await asyncio.sleep(seconds)
    print(title, "완료!")

async def main():
    start = time.perf_counter()
    await asyncio.gather(
        async_work("A작업",2),
        async_work("B작업",3),
        async_work("C작업",1)
        )

    end = time.perf_counter()

    print("걸린 시간", end - start)


await main()
```

##실습 과제

```python
import time
import asyncio

async def fetch_weather(location, t):

    print(f"{location} 날씨 데이터 수집 시작")
    await asyncio.sleep(t)
    print(f"{location} 날씨 데이터 수집 완료!")

async def main():
    start_time = time.perf_counter()

    await asyncio.gather(
            fetch_weather("서울", 1.5),
            fetch_weather("뉴욕", 2.5),
            fetch_weather("런던", 1.0))
    end_time = time.perf_counter()

    print(f"걸린 시간 {end_time - start_time:.1f}초")

await main()
```

```python
import asyncio

async def download_music(music_name, wait_seconds):

    print(f"{music_name} 다운로드를 시작합니다.")
    await asyncio.sleep(wait_seconds)
    print(f"{music_name} 다운로드가 완료되었습니다!")

async def main():

    await asyncio.gather(
            download_music("첫번째 음악", 2),
            download_music("두번째 음악", 4))


asyncio.run(main())
```

```python
await main()
```

```python
import asyncio
import time

async def send_notification(customer, seconds):
    print(f"{customer} 문자를 발송합니다.")
    await asyncio.sleep(seconds)
    print(f"{customer} 문자가 전송되었습니다.")

async def main():
    start_time = time.perf_counter()

    await asyncio.gather(
        send_notification("고객A", 1),
        send_notification("고객B", 3),
        send_notification("고객C", 2)
    )


    end_time = time.perf_counter()
    print(f"총 소요 시간: {end_time-start_time:.2f}초")

asyncio.run(main())
```

```python
await main()
```

```python
import datetime  # 날짜와 시각을 다루는 모듈
import time      # 소요 시간 측정 및 일시 정지 기능을 제공하는 모듈

# 현재 날짜와 시각을 구하고 원하는 포맷으로 출력합니다.
now = datetime.datetime.now()
print("1. 현재 시각 기록하기")
print(f"기본 출력: {now}")
print(f"포맷팅 출력: {now.strftime('%Y년 %m월 %d일 %H시 %M분 %S초')}")

# 프로그램이 가동되어 완료되기까지 몇 초가 소요되는지 계산합니다.
start_time = time.perf_counter()  # 타이머 측정 시작
time.sleep(0.5)                  # 의도적으로 0.5초 동안 대기
end_time = time.perf_counter()    # 타이머 측정 종료

elapsed = end_time - start_time
print("\n2. 소요 시간 측정하기")
print(f"대기 후 소요된 시간: {elapsed:.2f}초")
```
