2. 클래스(Class)와 인스턴스(Instance)

```python
class Bus:

    # 속성 정의
    def __init__(self):
        self.count = 0 # 승객 수, 초기화: 변수가 되었을때 최초로 값을 대입
        self.passengers = []

    def ride(self, passenger): # 탑승
        self.count += 1 # 승객 수 증가
        self.passengers.append(passenger)
        print("탑승")


    def off(self, passenger): # 하차
        if self.count > 1:
            self.count -= 1 # 승객 수 감소
            self.passengers.remove(self.passenger)

        print("하차")

    def show_count(self):
        print("승객 수:", self.count)

    def print_self(self):
        print(id(self))


class Passenger:
    pass
```

```python
bus1 = Bus()   # 객체, 클래스 --> 객체(인스턴스 - 실체)

# 마침표(.) 연산자를 통해서 객체의 변수, 메서드에 접근 가능

print("탑승객 수:", bus1.count)

bus1.ride()
```

```python
bus2= Bus()

bus1 is bus2 #같은 클래스로부터 생성된 객체는 서로 다른 객체!
```

```python
bus1 = Bus()
print("bus1 주소:", id(bus1))

bus1.print_self()
```

```python
bus = Bus()
p1 = Passenger()
p2 = Passenger()

bus.ride(p1)
bus.ride(p2)

bus.show_count()
```
