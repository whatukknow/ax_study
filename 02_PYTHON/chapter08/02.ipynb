1. 상속(Inheritance)
클래스 기반의 상속(상속에 대한 정의 -> 객체 생성(상속 관계 성립))
class 자식클래스(부모클래스):
    ... 
super: 부모 클래스를 가리키는 참조 변수

참고) self: 자식 객체(현재 객체)의 참조 변수

super() : 부모쪽 객체 생성

부모클래스 : 일반적인 정의들, 상위 개념

자식클래스 : 구제적인 정의들, 하위 개념

```python
class Animal:

    def __init__(self, kind):
        self.kind = kind # 동물 종류

    def move(self):
        print("동물이 움직인다.")


class Dog(Animal):
    def __init__(self):
        super().__init__("강아지") # 부모쪽 인스턴스 변수인 kind에 초기화
    
    

class Cat(Animal):
    def __init__(self):
        super().__init__("고양이")
```

```python
dog = Dog()
dog.move()

cat = Cat()
cat.move()
```

```python
print(dog.kind, cat.kind)
```

2. 메서드 오버라이딩(Overriding)
- 메서드 재정의
- 부모쪽 클래스와 동일한 이름으로 정의하여 기능을 수정, 재정의
- 자식쪽 클래스에 정의된 동일한 이름의 함수가 호출 우선순위를 가진다.
- 자식쪽에서 부모쪽 동일한 이름의 메서드를 호출해야하는 경우는 self로 접근 x
    - super(), 부모쪽 메서드(..)




```python
class Animal:

    def __init__(self, kind):
        self.kind = kind # 동물 종류

    def move(self):
        print("동물이 움직인다.")


class Dog(Animal):
    def __init__(self):
        super().__init__("강아지") # 부모쪽 인스턴스 변수인 kind에 초기화
    
    def move(self):
        super().move() #부모쪽 객체의 move()
        print("강아지 걸음으로 움직인다.")
    

class Cat(Animal):
    def __init__(self):
        super().__init__("고양이")

    def move(self):
        print("고양이 걸음으로 움직인다.")
```

```python
dog = Dog()
dog.move()

cat = Cat()
cat.move()
```

```python
class Human:
    def move(self):
        print("두발로 직립보행한다.")


human = Human()
human.move()
```

## 다중 상속 - 여러 부모클래스를 상속 관계로 정의 가능

다중 상속시 같은 메서드가 있을때 호출 우선순위가 있음. (MRO)
자식 클래스의 메서드 -> 상속 관계에서 왼쪽에 있는 클래스 -> 그 다음 오른쪽 클래스

```python
class ParentA:
    def a(self):
        print("ParentA - a()")

    def b(self):
        print("ParentA - b()")


class ParentB:
    def a(self):
        print("ParentB - a()")

    def c(self):
        print("ParentB - c()")


##class Child(ParentA, ParentB):
class Child(ParentB, ParentA):

    pass
```

```python
c = Child()
c.b()
c.c()
c.a()

```

```python
Child.__mro__ # method Resolution Order - 메서드 우선순위
```

```python
class Schedule:
    def __init__(self, year, month, day):
        self.year = year
        self.month = month
        self.day = day

    def __str__(self):
        return f"year={self.year}, month={self.month}, day={self.day}"
    
```

5. 파이썬 프라이빗(Private) 변수와 이름 변형 규칙 (Name Mangling)

- __변수명: 클래스 내부에서만 접근 가능한 인스턴스 변수가 된다.

```python
class Schedule:
    def __init__(self, year, month, day):
        self.__year = year
        self.__month = month
        self.__day = day


    def __str__(self):
        return f"year={self.__year}, month={self.__month}, day={self.__day}"
```

```python
s1 = Schedule(2026,8,18)

print(s1)
```

```python
s1.__year=2026
s1.__month=2
s1.__day=31

print(s1)
```

```python
class Schedule:
    def __init__(self, year, month, day):
        self.__year = year
        self.__month = month
        self.__day = day

    def get_year(self):
        return self.__year

    def get_month(self):
        return self.__month

    def get_day(self):
        return self.__day

    def set_year(self, year):
        self.__year = year

    def set_month(self, month):
        self.__month = month

    def set_day(self, day):
        if self.__month == 2 and day > 28:  #값에 대한 통제가 가능
            day = 28

        self.__day = day

    def __str__(self):
        return f"year={self.__year}, month={self.__month}, day={self.__day}"
```

```python
s1 = Schedule(2026,8,18)

s1.set_month(2)
s1.set_day(31)

print(s1)
```

```python
class Hero:
    def __init__(self, name):
        self.name = name
        self.hp = 100

    
    def take_damage(self, damage):
        self.hp -= damage
        print(f"{self.name}의 남은 체력: {self.hp}")

        if  self.hp < 0 :
            print(f"{self.name}님의 캐릭터가 쓰러졌습니다")


```

```python
hero = Hero("용사")

hero.take_damage(30)
hero.take_damage(40)
hero.take_damage(35)
```

```python
class Phone:
    def call(self):

        print("통화를 시작합니다")

class SmartPhone(Phone):
    def call(self):
        print("영상 통화를 시작합니다")

    def search_web(self):
        print("인터넷 검색을 시작합니다")

smartphone = SmartPhone()

smartphone.call()
smartphone.search_web()
```
