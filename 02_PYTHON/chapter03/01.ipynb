```python
2. 리스트(List)
- 변경 가능
- [값1, 값2, ...]

## 주요 기능
- append(요소): 끝에 추가
- insert(인덱스, 요소): 특정 위치에 추가
- extend(콜렉션 - 리스트, 튜플): 원본 리스트 끝에 여러개 추가
- [기존 리스트] + [값1, 값2, ...]: 기존 리스트에 새로운 리스트를 합쳐서 새로운 리스트 생성

조회(Read)
- 인덱스 번호로 접근 (fruits[1])
- index(요소): 특정 요소의 인덱스 번호

수정(Update)
- 인덱스 번호로 접근, 다른 값을 대입(fruits[1] = '오렌지')

삭제(Delete)
- remove(요소): 첫번째 발견한 특정 요소를 삭제
- clear(): 리스트의 모든 요소를 제거
- pop([인덱스]): 특정 위치의 요소를 삭제하고 반환
- del: 특정 위치의 요소를 삭제


기타
- sort() : 정렬(오름차순), sort(reverse=True) : 내림차순 정렬
- reverse() : 순서 반전
- count(...) : 특정 요소에 등장 횟수

```
