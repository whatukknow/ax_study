1. 딕셔너리(Dictionary) 자료 구조

- key - Value 발상의 구조
- key : Value를 찾기 위한 키워드, 중복 X, 리스트는 사용 불가
- Value: Key로 찾는 값, 중복 O
- {"키" : 값}

- 추가(Create): 없는 키이름으로 값을 대입하면 추가
- 조회(Read): ['키 이름'] - 키가 없으면 KeyError가 발생
  - keys(): 모든 키를 담은 View, 읽기 전용
  - values(): 모든 값을 담은 View, 읽기 전용
  - items(): (키, 값) 목록을 담은 View

```python
a = {10, 20, 30, 40}

b = {30, 40, 50, 60}

print("합집합:", a | b)

print("교집합:", a.intersection(b))

print("차집합:", a - b)

print("대칭차집합:", a^b)
```
