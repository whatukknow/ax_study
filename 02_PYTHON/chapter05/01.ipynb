```python
def add(*nums):
    #print(nums, type(nums))
    total = 0
    for num in nums:
        total += num

    return total
```

```python
import sys

def outer_func():
    x = 10
    
    def inner_func():
        y = 20
        
        # 1. 현재 실행 중인 프레임 객체 가져오기
        frame = sys._getframe()
        
        print(f"1. 현재 실행 함수: {frame.f_code.co_name}")
        print(f"2. 현재 지역 변수(f_locals): {frame.f_locals}")
        print(f"3. 전역 스코프 참조 여부: {'outer_func' in frame.f_globals}")
        
        # 2. f_back을 통해 부모 프레임(outer_func)으로 이동
        parent_frame = frame.f_back
        print(f"4. 부모 함수 이름: {parent_frame.f_code.co_name}")
        print(f"5. 부모 함수의 지역 변수: {parent_frame.f_locals}")

    inner_func()

outer_func()
```

```python
def get_total_and_average(*score):
    if len(score) ==0:
     return 0, 0

    total = sum(score)
    average = total / len(score)
    return total, average

total1, average1 = get_total_and_average(90,85,95)
print(total1, average1)

total2, average2 = get_total_and_average(70,80,90,100,65)
print(total2, average2)
```
