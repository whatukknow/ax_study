2. 파이썬의 변수 검색 규칙: LEGB 스코프
접근 우선순위: 좁은 범위가 우선순위가 높다!
L(Local - 지역 변수) - E(Enclosing - 외부 함수 변수) - G(Global - 전역변수) - B(Builtin- 내장 변수)


```python
def outer():
    a = 'Enclosing'
    print("outer:", a)

    def inner():
        a = 'Local'
        print("inner:", a)
    inner()

outer()
```

```python
a = 'Global'
def outer():
	a = 'Enclosing'
	print('outer:', a)

	def inner():
			global a  #a라는 변수는 전역변수를 찾겠다.
			print("inner:",a)
	
	inner()

outer()
```

```python
a = 'Global'
def outer():
	a = 'Enclosing'
	print('outer:', a)

	def inner():
			global a  #a라는 변수는 전역변수를 찾겠다.
			a = not a
			print("inner:",a)
	
	inner()

outer()
```

```python
def callback():
    print("callback()")

def func(a):
    print("func()")
    a()

func(callback)
```

```python
def outer(num1):
    def inner(num2):
        return num1 + num2
    return inner
```

```python
a = outer(10)   #항상 10을 더해주는 새로운 함수 - num 1이 inner 함수에 속박(closure)
a(20) # 10+20

b = outer(20)
b(20) # 10+30
```

```python
a(30)  # 10 + 30
```

```python
b = outer(20)  #항상 20을 더해주는 새로운 함수
b(20)  # 20 + 20
```

```python
def change_numbers(*args, callback):
    numbers = list(args)

    for i in range(len(numbers)):
        numbers[i] = callback(numbers[i])

    return numbers


def square(num):
    return num ** 2


change_numbers(10,20,30, callback=square)
```

```python
def plus(num):
    return num + 10

change_numbers(10,20,30, callback=plus)
```

```python
def change_numbers(callback, *args):
    numbers = list(args)

    for i in range(len(numbers)):
        numbers[i] = callback(numbers[i])

    return numbers
```

```python
change_numbers(lambda num:num**2, 10,20,30)
```
