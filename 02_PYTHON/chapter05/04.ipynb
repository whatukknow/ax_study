3. 제너레이터(Generator)

```python
def generate_nums():

    print("1번째")
    yield 10

    print("2번째")
    yield 20

    print("3번재")
    yield 30
```

```python
gen = generate_nums()
gen
```

```python
next(gen)
```

4. 제너레이터 표현식

```python
song_list=['놈','해','달']

play_list=iter(song_list)
```

```python
def generate_keyword():

    print("1번째")
    yield "대박"

    print("2번째")
    yield "사건"

    print("3번재")
    yield "성공"
```

```python
gen = generate_keyword()
print(next(gen))
print(next(gen))
print(next(gen))
```

```python
print(next(play_list))
print(next(play_list))
print(next(play_list))

#stopiteration
print(next(play_list))
```
