```python
nums =[*range(1,11)]

nums2 = [x**2 for x in nums]

print(nums2)

nums3 = [x for x in nums if x % 2 == 1]
print(nums3)
```

Zip()

참고 - itertools 모듈 zip_longest


```python
persons = ["김철수", "김영희", "김민수"]
fruits = ["사과", "오렌지", "멜론", "망고"]
```

```python
items = []
for i, person in enumerate(persons):
    items.append((person, fruits[i]))

items
```

```python
items = zip(persons, fruits)
list(items)
```

sorted() 함수

- 리스트.sort(): 원본 데이터를 정렬
- sorted(Iterable): 원본 데이터를 유지, 새로운 객체를 생성해서 정렬(불변 객체)

```python
import random

nums = [*range(1,11)]
random.shuffle(nums)

nums
```

```python
nums.sort()

nums
```

```python
nums_sorted = sorted(nums)

print("nums_sorted:", nums_sorted)
print("nums:", nums)
```

min(), max()

- min(): 최소값 / 오름 차순 정렬 후 가장 처음 요소
- max(): 최댓값 / 오름 차순 정렬 후 가장 마지막 요소

```python
min(), max()

- min(): 최소값
- max(): 최댓값
```

```python
import json

# 파이썬 객체 -> JSON 문자열

user = {"name": "김철수", "age": 20}

json_string = json.dumps(user, ensure_ascii=False)
print("json_string:", json_string, "type:", type(json_string))

```
